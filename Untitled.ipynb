```python
from sqlalchemy import create_engine
import pandas as pd

engine = create_engine('postgresql://postgres@localhost:5432/retail_sales')
```

```python
#CSVファイルの読み込みとテーブル書き込み
##Loading CSV Files and Writing to Tables
```

```python
df_customers = pd.read_csv('Customers.csv')
df_products = pd.read_csv('Products.csv')
df_stores = pd.read_csv('Stores.csv')
df_transactions = pd.read_csv('Transactions.csv')

df_customers.to_sql('customers', engine, if_exists='replace', index=False)
df_products.to_sql('products', engine, if_exists='replace', index=False)
df_stores.to_sql('stores', engine, if_exists='replace', index=False)
df_transactions.to_sql('transactions', engine, if_exists='replace', index=False)

print("4つのテーブルをすべて書き込みました!")
```

```python
#Profit by product and Category of the product per unit
#元のデータに通貨の記載はありませんので、ご了承ください
```

```python
%%sql
SELECT ROUND(p."UnitPrice" - p."CostPrice") AS "Profit", p."ProductName", p."Category" 
FROM products p
ORDER BY "Profit" DESC ;
```

```python
#Total profit by category/カテゴリ別の総利益
```

```python
%%sql
SELECT p."Category", to_char(SUM((p."UnitPrice" - p."CostPrice") * t."Quantity"),'FM999,999,999') AS "Total_profit_by_category"
FROM transactions t
JOIN products p ON t."ProductID" = p."ProductID"
GROUP BY p."Category";
```

```python
#Total profit by region/地域別の利益
```

```python
%%sql
SELECT s."Region", to_char(SUM((p."UnitPrice" - p."CostPrice") * t."Quantity"), 'FM999,999,999') AS "Total_profit_by_region"
FROM transactions t
JOIN products p ON t."ProductID" = p."ProductID"
JOIN stores s ON t."StoreID" = s."StoreID"
GROUP BY s."Region"
ORDER BY SUM((p."UnitPrice" - p."CostPrice") * t."Quantity") DESC;
```

```python
#利益率(Profit Margin)ランキング/Profit Margin Ranking
#ROUND( (利益 / 単価) * 100, 1 )で利益率をパーセンテージに変換、最後に小数点１桁に変更
#ヌメリック型に::numericを使って、計算結果を明示的にnumeric型に変換してからROUND
#==========English below===========
#Use `ROUND((Profit / Unit Price) * 100, 1)` to convert the profit margin to a percentage, then round to one decimal place
#Use `::numeric` to explicitly convert the calculation result to the numeric type before applying `ROUND`
```

```python
%%sql
SELECT p."ProductName", p."Category",
    ROUND((((p."UnitPrice" - p."CostPrice") / p."UnitPrice") * 100)::numeric, 1) AS "Profit_margin_percent"
FROM products p
ORDER BY "Profit_margin_percent" DESC
LIMIT 10;
```

```python
#割引の影響分析/Analysis of the Impact of Discounts
#1行目:割引が適用された取引だけを集めた時の、合計利益
#2行目:割引なしで売れた取引だけを集めた時の、合計利益
#==========English below===========
#Row 1: Total profit from transactions where a discount was applied
#Row 2: Total profit from transactions sold without a discount
```

```python
%%sql
SELECT
    CASE WHEN t."Discount" > 0 THEN 'Discounted' ELSE 'Full Price' END AS "Discount_status",
    to_char(SUM((p."UnitPrice" - p."CostPrice") * t."Quantity"), 'FM999,999,999') AS "Total_profit"
FROM transactions t
JOIN products p ON t."ProductID" = p."ProductID"
GROUP BY "Discount_status";
```

```python
#平均利益(1件あたり)で比較する
#==========English below===========
#Comparing by Average Profit (Per Transaction)
```

```python
%%sql
SELECT
    CASE WHEN t."Discount" > 0 THEN 'Discounted' ELSE 'Full Price' END AS "Discount_status",
    COUNT(*) AS "Number_of_transactions",
    to_char(AVG((p."UnitPrice" - p."CostPrice") * t."Quantity"), 'FM999,999,999') AS "Average_profit_per_transaction"
FROM transactions t
JOIN products p ON t."ProductID" = p."ProductID"
GROUP BY "Discount_status";
```

```python
#上記の結果から、下記の結果がわかる。
#割引ありの合計利益が大きかったのは、単純に件数が3倍多かったからです。
```

```python
#顧客セグメント(性別)別の利益及びトランザクション件数、１件あたりの平均利益
#===========English below===============
#Profit and Number of Transactions by Customer Segment (Gender), and Average Profit Per Transaction
```

```python
%%sql
SELECT c."Gender",
    COUNT(*) AS "Number_of_transactions",
    to_char(SUM((p."UnitPrice" - p."CostPrice") * t."Quantity"), 'FM999,999,999') AS "Total_profit",
    to_char(AVG((p."UnitPrice" - p."CostPrice") * t."Quantity"), 'FM999,999,999') AS "Average_profit_per_transaction"
FROM transactions t
JOIN products p ON t."ProductID" = p."ProductID"
JOIN customers c ON t."CustomerID" = c."CustomerID"
GROUP BY c."Gender";
```

```python
#上記の結果から、件数の差だけでなく、1件あたりの効率もわずかに男性の方が高い
```

```python
#月次の利益トレンド/Monthly Profit Trends
#DATE_TRUNC('month', t."Date"):日付を「月単位」に切り詰めます。
#ORDER BY "Month"で、時系列順(古い月から新しい月へ)に並べます。
# ::dateを使って、文字列を日付型に変換する
#==================English below===================
# DATE_TRUNC(‘month’, t.“Date”): Truncates the date to the “month” level.
# ORDER BY “Month” sorts the data in chronological order (from oldest to newest month).
# Uses ::date to convert a string to a date type
```

```python
%%sql
SELECT DATE_TRUNC('month', t."Date"::date) AS "Month",
    COUNT(*) AS "Number_of_transactions",
    to_char(SUM((p."UnitPrice" - p."CostPrice") * t."Quantity"), 'FM999,999,999') AS "Monthly_profit"
FROM transactions t
JOIN products p ON t."ProductID" = p."ProductID"
GROUP BY DATE_TRUNC('month', t."Date"::date)
ORDER BY "Month";
```

```python
#年代別の分析/Analysis by Age Group
#まず年齢を計算し、年代(10歳刻み)にグループ分け
#AGE(CURRENT_DATE, c."BirthDate"::date) - 今日の日付と生年月日の差を計算し、年齢を出します
#EXTRACT(YEAR FROM ...):その差から「年」の部分だけを取り出します(例:「32年3ヶ月」から「32」だけ取る)
#CASE WHENで、年齢を10歳刻みの年代グループに振り分けます
#==================English below===================
#First, calculate age and group by age bracket (in 10-year increments)
#AGE(CURRENT_DATE, c.“BirthDate”::date) - Calculate the difference between today's date and the birthdate to determine age
#EXTRACT(YEAR FROM ...): Extract only the “years” portion from that difference (e.g., extract only “32” from “32 years and 3 months”)
#Use CASE WHEN to assign ages to age groups in 10-year increments
```

```python
%%sql
SELECT
    CASE 
        WHEN EXTRACT(YEAR FROM AGE(CURRENT_DATE, c."BirthDate"::date)) < 20 THEN '10代'
        WHEN EXTRACT(YEAR FROM AGE(CURRENT_DATE, c."BirthDate"::date)) < 30 THEN '20代'
        WHEN EXTRACT(YEAR FROM AGE(CURRENT_DATE, c."BirthDate"::date)) < 40 THEN '30代'
        WHEN EXTRACT(YEAR FROM AGE(CURRENT_DATE, c."BirthDate"::date)) < 50 THEN '40代'
        WHEN EXTRACT(YEAR FROM AGE(CURRENT_DATE, c."BirthDate"::date)) < 60 THEN '50代'
        ELSE '60代以上'
    END AS "Age_group",
    COUNT(*) AS "Number_of_transactions",
    to_char(SUM((p."UnitPrice" - p."CostPrice") * t."Quantity"), 'FM999,999,999') AS "Total_profit",
    to_char(AVG((p."UnitPrice" - p."CostPrice") * t."Quantity"), 'FM999,999,999') AS "Average_profit_per_transaction"
FROM transactions t
JOIN products p ON t."ProductID" = p."ProductID"
JOIN customers c ON t."CustomerID" = c."CustomerID"
GROUP BY "Age_group"
ORDER BY "Age_group";
```

```python
#簡単な考察/Brief Analysis
#取引件数は60代以上(1,285件)が最多、次いで20代(1,074件)
#1件あたりの平均利益は10代(1,404)が最も高いですが、件数がわずか19件しかないため、これはサンプル数が少なすぎて信頼性が低い結果です(1〜2件の高額取引で平均が引っ張られている可能性が高いです)
#実質的に比較に値するグループ(20代〜60代以上)では、平均利益は961〜1,036の範囲に収まっており、年代による大きな差はあまり見られないという結果です。
#==================English below===================
#The number of transactions was highest among those in their 60s and older (1,285), followed by those in their 20s (1,074).
#The average profit per transaction was highest among those in their teens (1,404), 
#but since there were only 19 transactions in this group, the sample size is too small for the results to be reliable 
#(it is highly likely that the average is skewed by one or two high-value transactions).
#Among the groups that are practically comparable (those in their 20s through those aged 60 and older), 
#the average profit falls within the range of 961 to 1,036, indicating that there are no significant differences based on age.
```

```python
#利益貢献度が高い顧客トップ10/Top 10 Customers by Profit Contribution
#顧客ごと(CustomerID単位)に、利益を合計します
#ORDER BY ... DESCで利益が高い順に並べ、LIMIT 10で上位10人だけに絞ります
#1件あたりの平均利益
#==================English below===================
#Sum the profit for each customer (by Customer ID)
#Use `ORDER BY ... DESC` to sort by profit in descending order, and `LIMIT 10` to filter the top 10 customers
#Average profit per customer
```

```python
%%sql
SELECT c."FirstName", c."LastName", c."CustomerID",
    COUNT(*) AS "Number_of_transactions",
    to_char(SUM((p."UnitPrice" - p."CostPrice") * t."Quantity"), 'FM999,999,999') AS "Total_profit",
    to_char(AVG((p."UnitPrice" - p."CostPrice") * t."Quantity"), 'FM999,999,999') AS "Average_profit_per_transaction"
FROM transactions t
JOIN products p ON t."ProductID" = p."ProductID"
JOIN customers c ON t."CustomerID" = c."CustomerID"
GROUP BY c."CustomerID", c."FirstName", c."LastName"
ORDER BY SUM((p."UnitPrice" - p."CostPrice") * t."Quantity") DESC
LIMIT 10;
```

```python
#上記の結果からわかることは下記に記載しました。/summarized the key takeaways from the above results below.
#上位顧客の利益は、34,000〜41,000の範囲に集中しています。
#トップ(Anna Andrews:41,190)と10位(Jenny Beck:34,821)の差はそこまで大きくなく、特定の1人だけが突出しているわけではありません。
#取引件数と利益額は、必ずしも比例していないことがわかります。
#Number_of_transactions(件数)が多いのに、Average_profit_per_transaction(平均)が低い顧客 → 「常連客タイプ」(例：Vicki Guzman)
#Number_of_transactionsが少ないのに、Average_profit_per_transactionが高い顧客 → 「高単価顧客タイプ」(例：Anna Andrews)
#==================English below===================
#Profits from top customers are concentrated in the range of 34,000 to 41,000.
#The difference between the top customer (Anna Andrews: 41,190) and the 10th-ranked customer (Jenny Beck: 34,821) is not that significant, so it’s not the case that just one specific person stands out.
#It is clear that the number of transactions and profit amount are not necessarily proportional.
#Customers with a high Number_of_transactions but a low Average_profit_per_transaction → “Regular Customer Type” (e.g., Vicki Guzman)
#Customers with a low Number_of_transactions but a high Average_profit_per_transaction → “High-Value Customer Type” (e.g., Anna Andrews)
```

```python
#支払い方法別の分析/ Payment method analysis
#PaymentMethod(クレジットカード、現金、電子マネーなど)ごとにグループ分けします
#これまでと同じく、件数・合計利益・平均利益の3つを一緒に見られるようにしています
#============== English below===================
#Data is grouped by payment method (credit card, cash, electronic money, etc.)
#As before, the data is organized so you can view the number of transactions, total profit, and average profit all at once
```

```python
%%sql
SELECT t."PaymentMethod",
    COUNT(*) AS "Number_of_transactions",
    to_char(SUM((p."UnitPrice" - p."CostPrice") * t."Quantity"), 'FM999,999,999') AS "Total_profit",
    to_char(AVG((p."UnitPrice" - p."CostPrice") * t."Quantity"), 'FM999,999,999') AS "Average_profit_per_transaction"
FROM transactions t
JOIN products p ON t."ProductID" = p."ProductID"
GROUP BY t."PaymentMethod"
ORDER BY SUM((p."UnitPrice" - p."CostPrice") * t."Quantity") DESC;
```

```python
#分析結果/Analysis result
#取引件数はほぼ均等(1,159〜1,284件)に分散しています。これは、顧客が特定の支払い方法に極端に偏っていない、バランスの取れたデータであることを示しています
#1件あたりの平均利益で見ると、Bank Transfer(銀行振込)が最も高い(1,027)一方、件数は最も少ない(1,159件)という結果になっています
#差は全体的に小さい(978〜1,027の範囲)ので、これも先ほどの性別分析と同様、「支払い方法によって、利益に決定的な差があるとは言えない」というのが正直な結論です。
#ただし、Bank Transferがわずかに高い傾向は、興味を持って良い切り口です。
#============== English below===================
#The number of transactions is distributed fairly evenly (ranging from 1,159 to 1,284). This indicates that the data is well-balanced, with customers not showing an extreme bias toward any specific payment method.
#In terms of average profit per transaction, Bank Transfer yields the highest figure (1,027), while the number of transactions is the lowest (1,159).
#Since the differences are generally small (ranging from 978 to 1,027), the honest conclusion—similar to the gender analysis mentioned earlier—is that “it cannot be said that there is a decisive difference in profit based on payment method.”
#However, the slight tendency for Bank Transfer to yield higher profits is an interesting angle worth exploring.
```

```python
#考えられる仮説/ Prediction
#銀行振込を使う客は、高額商品を購入する際に選ばれやすい支払い方法である可能性がある
#Bank transfer may be a payment method that customers are more likely to choose when purchasing high-priced items.
```
